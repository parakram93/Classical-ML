# Evaluation Metrics — Notes

## 🔹 Regression Metrics

### 1. MAE (Mean Absolute Error)
$$MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|$$

- Average of absolute errors.
- Same unit as target → easy to interpret.
- Treats all errors equally → robust to outliers.

**Example:** Actual = [10, 20, 30], Predicted = [12, 18, 33]
Errors = [2, 2, 3] → **MAE = 2.33**

---

### 2. MSE (Mean Squared Error)
$$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

- Average of squared errors.
- Penalizes large errors heavily.
- Unit = (target unit)² → less interpretable.
- Sensitive to outliers.

**Example (same data):** Squared errors = [4, 4, 9] → **MSE = 5.67**

---

### 3. RMSE (Root Mean Squared Error)
$$RMSE = \sqrt{MSE}$$

- Square root of MSE → brings back to original unit.
- Still penalizes large errors more than MAE.
- Most commonly reported regression metric.

**Example:** RMSE = √5.67 = **2.38**

> 🧠 **Rule of thumb:** RMSE ≥ MAE always. If RMSE ≫ MAE → model has a few large errors (outliers).

---

### 4. R² (Coefficient of Determination)
$$R^2 = 1 - \frac{SS_{residual}}{SS_{total}} = 1 - \frac{\sum(y_i-\hat{y}_i)^2}{\sum(y_i-\bar{y})^2}$$

- Measures variance explained by the model, vs. just predicting the mean.
- **R² = 1** → perfect model
- **R² = 0** → same as predicting the mean
- **R² < 0** → model worse than predicting the mean

⚠️ R² tends to increase as you add more features (even useless ones) → use **Adjusted R²** to penalize that.

---

### 📌 Regression Summary Table

| Metric | Formula Basis | Interpretability | Outlier Sensitivity |
|--------|---------------|-------------------|----------------------|
| MAE  | Absolute error | Same unit as target | Low |
| MSE  | Squared error | Unit² (hard to read) | High |
| RMSE | √MSE | Same unit as target | High |
| R²   | Variance explained | 0–1 scale (relative) | Moderate |

## 🔹 Classification Metrics

### Confusion Matrix (Binary)

|                     | Predicted Positive | Predicted Negative |
|---------------------|---------------------|----------------------|
| **Actual Positive** | TP (True Positive)  | FN (False Negative) |
| **Actual Negative** | FP (False Positive) | TN (True Negative)  |

- **FP** → Type I error ("false alarm")
- **FN** → Type II error ("missed detection")

**Example (100 samples):**

|                    | Pred: Positive | Pred: Negative |
|--------------------|-----------------|-----------------|
| **Actual: Positive** | 40 (TP)       | 10 (FN)         |
| **Actual: Negative** | 5 (FP)        | 45 (TN)         |

---

### 1. Accuracy
$$Accuracy = \frac{TP+TN}{TP+TN+FP+FN}$$

= (40+45)/100 = **0.85**

⚠️ Misleading on **imbalanced datasets** (e.g., predicting the majority class always still gives high accuracy).

---

### 2. Precision
$$Precision = \frac{TP}{TP+FP}$$

= 40 / (40+5) = **0.889**

- "When model predicts positive, how often is it right?"
- Matters when **false positives are costly**.

---

### 3. Recall (Sensitivity)
$$Recall = \frac{TP}{TP+FN}$$

= 40 / (40+10) = **0.80**

- "Of all actual positives, how many did we catch?"
- Matters when **false negatives are costly**.

---

### 4. F1 Score
$$F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}$$

= 2 × (0.889 × 0.80)/(0.889+0.80) = **0.842**

- Harmonic mean of Precision & Recall.
- Punishes imbalance between the two (better than simple average).
- Good single metric for **imbalanced classes**.

---

### 📌 Classification Summary Table

| Metric | Formula | Best Used When |
|--------|---------|------------------|
| Accuracy  | (TP+TN)/Total | Classes are balanced |
| Precision | TP/(TP+FP) | False positives are costly |
| Recall    | TP/(TP+FN) | False negatives are costly |
| F1 Score  | Harmonic mean of P & R | Imbalanced classes, need 1 balanced metric |

---

## 🧭 Quick Decision Guide

| Situation | Preferred Metric(s) |
|-----------|----------------------|
| Regression, interpretable & robust | MAE |
| Regression, penalize large errors | RMSE |
| Regression, explain variance | R² / Adjusted R² |
| Balanced classes | Accuracy |
| Costly false alarms | Precision |
| Costly missed detections | Recall |
| Imbalanced classes, need one score | F1 |

> 💡 For imbalanced classification tasks (e.g., hallucination detection where hallucinated samples are the minority class), rely on **Precision, Recall, F1, and the full Confusion Matrix** rather than Accuracy alone.
